```python
# Bubble Sort

import random

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

n = int(input("Enter the size of array: "))
arr = [random.randint(0, 100) for i in range(n)]
print("Original Array: ", arr)
bubble_sort(arr)
print("Sorted Array:", arr)
```

```python
# Insertion Sort

import random

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

n = int(input("Enter the size of array: "))
arr = [random.randint(0, 100) for i in range(n)]
print("Original Array: ", arr)
insertion_sort(arr)
print("Sorted Array:", arr)
```

```python
# Merge Sort

import random

def merge(arr, left, mid, right):
    n1 = mid - left + 1
    n2 = right - mid

    left_part = [0] * (n1 + 1)
    right_part = [0] * (n2 + 1)

    for i in range(n1):
        left_part[i] = arr[left + i]

    for j in range(n2):
        right_part[j] = arr[mid + j + 1]

    left_part[n1] = float('inf')
    right_part[n2] = float('inf')

    i = 0
    j = 0

    for k in range(left, right + 1):
        if left_part[i] <= right_part[j]:
            arr[k] = left_part[i]
            i += 1
        else:
            arr[k] = right_part[j]
            j += 1

def merge_sort(arr, left, right):
    if left < right:
        mid = (left + right) // 2
        merge_sort(arr, left, mid)
        merge_sort(arr, mid + 1, right)
        merge(arr, left, mid, right)


n = int(input("Enter the size of array: "))
arr = [random.randint(0, 100) for i in range(n)]
print("Original Array: ", arr)
merge_sort(arr, 0, n - 1)
print("Sorted Array:", arr)
```

```python
# Quick Sort

import random

def partition(array, start, end):
    pivot = array[end]
    i = start - 1
    for j in range(start, end):
        if array[j] <= pivot:
            i += 1
            array[i], array[j] = array[j], array[i]
    array[i + 1], array[end] = array[end], array[i + 1]
    return i + 1

def quick_sort(array, start, end):
    if start < end:
        q = partition(array, start, end)
        quick_sort(array, start, q - 1)
        quick_sort(array, q + 1, end)

n = int(input("Enter the size of array: "))
arr = [random.randint(0, 100) for i in range(n)]
print("Original Array: ", arr)
quick_sort(arr, 0, n - 1)
print("Sorted Array:", arr)
```

```python
# Heap Sort

import random

def left(i):
    return 2 * i + 1

def right(i):
    return 2 * i + 2

def max_heapify(arr, size, i):
    largest = i
    l = left(i)
    r = right(i)

    if l < size and arr[l] > arr[largest]:
        largest = l
    if r < size and arr[r] > arr[largest]:
        largest = r
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        max_heapify(arr, size, largest)

def build_max_heap(arr):
    size = len(arr)
    for i in range(size // 2 - 1, -1, -1):
        max_heapify(arr, size, i)

def heap_sort(arr):
    size = len(arr)
    build_max_heap(arr)
    for i in range(size - 1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]
        max_heapify(arr, i, 0)

n = int(input("Enter the size of array: "))
arr = [random.randint(0, 100) for _ in range(n)]
print("Original Array:", arr)
heap_sort(arr)
print("Sorted Array:", arr)
```

```python
# Priority Queue

def left(i):
    return 2 * i + 1

def right(i):
    return 2 * i + 2

def parent(i):
    return (i - 1) // 2

def max_heapify(arr, i):
    largest = i
    l = left(i)
    r = right(i)
    size = len(arr)

    if l < size and arr[l] > arr[largest]:
        largest = l
    if r < size and arr[r] > arr[largest]:
        largest = r
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        max_heapify(arr, largest)

def heap_maximum(arr):
    if len(arr) > 0:
        return arr[0]
    else:
        return None

def heap_extract_max(arr):
    if len(arr) < 1:
        raise IndexError("heap underflow")
    max_val = arr[0]
    arr[0] = arr[-1]
    arr.pop()
    max_heapify(arr, 0)
    return max_val

def heap_increase_key(arr, i, key):
    if key < arr[i]:
        raise ValueError("new key is smaller than current key")
    arr[i] = key
    while i > 0 and arr[parent(i)] < arr[i]:
        p = parent(i)
        arr[i], arr[p] = arr[p], arr[i]
        i = p

def max_heap_insert(arr, key):
    arr.append(float('-inf'))
    heap_increase_key(arr, len(arr) - 1, key)

heap = []
print("\nPriority Queue Menu:")
print("1. Insert")
print("2. Get Maximum")
print("3. Extract Maximum")
print("4. Increase Key")
print("5. Show Heap")
print("6. Exit")

while True:
    ch = input("\nEnter your choice (1-6): ")
    if ch == "1":
        val = int(input("Enter value to insert: "))
        max_heap_insert(heap, val)
    elif ch == "2":
        print("Maximum value:", heap_maximum(heap))
    elif ch == "3":
        try:
            print("Extracted maximum:", heap_extract_max(heap))
        except IndexError as e:
            print(e)
    elif ch == "4":
        i = int(input("Enter index to increase: "))
        if 0 <= i < len(heap):
            new_key = int(input("Enter new key: "))
            try:
                heap_increase_key(heap, i, new_key)
            except ValueError as e:
                print(e)
        else:
            print("Invalid index")
    elif ch == "5":
        print("Heap:", heap)
    elif ch == "6":
        print("Exiting.")
        break
    else:
        print("Invalid choice. Try again.")
```

```python
# Counting Sort

import random

def counting_sort(A, B, k):
    C = [0] * (k + 1)

    for j in range(len(A)):
        C[A[j]] += 1

    for i in range(1, k + 1):
        C[i] += C[i - 1]

    for j in range(len(A) - 1, -1, -1):
        B[C[A[j]] - 1] = A[j]
        C[A[j]] -= 1

    return B

n = int(input("Enter the size of array: "))
A = [random.randint(0, 20) for i in range(n)]
B = [0] * n
k = max(A)

print("Original array:", A)
sorted_arr = counting_sort(A, B, k)
print("Sorted array:  ", sorted_arr)
```

```python
# Radix Sort

import random

def counting_sort(A, k, exp):
    B = [0] * len(A)
    C = [0] * (k + 1)

    for j in range(len(A)):
        index = (A[j] // exp) % 10
        C[index] += 1

    for i in range(1, k + 1):
        C[i] += C[i - 1]

    for j in range(len(A) - 1, -1, -1):
        index = (A[j] // exp) % 10
        B[C[index] - 1] = A[j]
        C[index] -= 1

    return B

def radix_sort(A, d):
    k = 9
    exp = 1

    for i in range(d):
        A = counting_sort(A, k, exp)
        exp *= 10

    return A

num = int(input("Enter the number of elements to sort: "))
A = [random.randint(0, 999) for i in range(num)]
print("Unsorted array:", A)

max_digits = len(str(max(A)))
sorted_A = radix_sort(A, max_digits)
print("Sorted array:  ", sorted_A)
```

```python
# Bucket Sort

import random

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

def bucket_sort(A):
    n = len(A)

    B = []
    for i in range(n):
        B.append([])

    for i in range(n):
        index = int(n * A[i])
        B[index].append(A[i])

    for i in range(n):
        insertion_sort(B[i])

    result = []
    for i in range(n):
        result += B[i]

    return result

n = int(input("Enter the number of elements: "))
A = [round(random.random(), 2) for i in range(n)]

print("Original array:", A)
sorted_A = bucket_sort(A)
print("Sorted array:  ", sorted_A)
```

```python
# BSF (Breadth-First Search)

def bfs(graph, start):
    visited = {}
    for i in graph:
        visited[i] = False
        
    queue = []
    
    visited[start] = True
    queue.append(start)
    
    print(f"BFS starting from vertex {start}:")
    
    while queue:
        u = queue.pop(0)
        print(u, end=" ")

        for w in graph[u]:
            if not visited[w]:
                queue.append(w)
                visited[w] = True
    print()

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

bfs(graph, 'A')
```

```python
# DFS (Depth-First Search)

def dfs(graph, v, visited):
    
    visited[v] = True
    print(v, end=' ')

    for i in graph[v]:
        if not visited[i]:
            dfs(graph, i, visited)

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

visited = {}
for i in graph:
    visited[i] = False

dfs(graph, 'A', visited)
```

```python
# Prim's Algorithm

def prim(cost, n):
    near = [0] * n
    t = []
    min_cost = 0

    min_val = float('inf')
    k = l = 0
    for i in range(n):
        for j in range(n):
            if cost[i][j] < min_val and i != j:
                min_val = cost[i][j]
                k, l = i, j

    t.append((k, l))
    min_cost += cost[k][l]

    for i in range(n):
        if cost[i][l] < cost[i][k]:
            near[i] = l
        else:
            near[i] = k

    near[k] = near[l] = -1

    for i in range(n - 2):
        min_val = float('inf')
        j = -1
        for i in range(n):
            if near[i] != -1 and cost[i][near[i]] < min_val:
                j = i
                min_val = cost[i][near[i]]

        t.append((j, near[j]))
        min_cost += cost[j][near[j]]
        near[j] = -1

        for k in range(n):
            if near[k] != -1 and cost[k][near[k]] > cost[k][j]:
                near[k] = j

    return t, min_cost

INF = float('inf')
cost_matrix = [
    [INF, 2,   INF, 6,   INF],
    [2,   INF, 3,   8,   5],
    [INF, 3,   INF, INF, 7],
    [6,   8,   INF, INF, 9],
    [INF, 5,   7,   9,   INF]
]

edges, total_cost = prim(cost_matrix, 5)

print("Edges in MST:", edges)
print("Total cost of MST:", total_cost)
```

```python
# Kruskal's Algorithm

def find(u, parent):
    while parent[u] != u:
        u = parent[u]
    return u

def union(u, v, parent):
    parent_u = find(u, parent)
    parent_v = find(v, parent)
    parent[parent_u] = parent_v
    
def kruskal(n, edge_list):
    parent = [i for i in range(n)]
    
    edge_list.sort(key=lambda x: x[2])
    
    t = []
    mincost = 0
    i = 0

    for u, v, cost in edge_list:
        if i == n - 1:
            break
        j = find(u, parent)
        k = find(v, parent)
        if j != k:
            t.append((u, v))
            mincost += cost
            union(j, k, parent)
            i += 1

    if i != n - 1:
        print("No spanning tree")
        return None, []
    else:
        return mincost, t

edges = [
    (0, 1, 10),
    (0, 2, 6),
    (0, 3, 5),
    (1, 3, 15),
    (2, 3, 4)
]

n = 4
mincost, mst = kruskal(n, edges)
print("Minimum cost:", mincost)
print("Edges in MST:", mst)
```

```python
# Dijkstra's Algorithm

def dijkstra(v, cost, n):
    INF = float('inf')
    S = [0] * n
    DIST = [INF] * n

    for i in range(n):
        S[i] = 0
        DIST[i] = cost[v][i]

    DIST[v] = 0
    S[v] = 1

    for _ in range(n):
        min_dist = INF
        u = -1
        for j in range(n):
            if S[j] == 0 and DIST[j] < min_dist:
                min_dist = DIST[j]
                u = j

        if u == -1:
            break

        S[u] = 1

        for w in range(n):
            if S[w] == 0 and cost[u][w] != INF:
                DIST[w] = min(DIST[w], DIST[u] + cost[u][w])

    return DIST

INF = float('inf')
cost_matrix = [
    [0, 10, INF, 30, 100],
    [INF, 0, 50, INF, INF],
    [INF, INF, 0, INF, 10],
    [INF, INF, 20, 0, 60],
    [INF, INF, INF, INF, 0]
]

source = 0
n = len(cost_matrix)
distances = dijkstra(source, cost_matrix, n)

print("Shortest distances from source:", distances)
```

```python
# Floyd-Warshall Algorithm

def floyd_warshall(cost):
    n = len(cost)
    A = [[cost[i][j] for j in range(n)] for i in range(n)]

    for k in range(n):
        for i in range(n):
            for j in range(n):
                A[i][j] = min(A[i][j], A[i][k] + A[k][j])

    return A

INF = float('inf')

cost_matrix = [
    [0,   3,   INF, 5],
    [2,   0,   INF, 4],
    [INF, 1,   0,   INF],
    [INF, INF, 2,   0]
]

shortest_paths = floyd_warshall(cost_matrix)

print("All-pairs shortest paths:")
for row in shortest_paths:
    print(row)
```

```python
# Recursive Activity Selector

def iterative_activity_selector(s, f):
    n = len(s) 
    A = [0]
    k = 0
    for i in range(1, n):
        if s[i] >= f[k]:
            A.append(i)
            k = i
    return A

s = [1, 3, 0, 5, 8, 5]
f = [2, 4, 6, 7, 9, 9]
selected_activities = iterative_activity_selector(s, f)
print("Selected activities (iterative):", selected_activities)
```

```python
# Iterative Activity Selector

def recursive_activity_selector(s, f, k, n):
    m = k + 1
    while m <= n and s[m] < f[k]:
        m += 1
    if m <= n:
        return [m] + recursive_activity_selector(s, f, m, n)
    else:
        return []

s = [1, 3, 0, 5, 8, 5]
f = [2, 4, 6, 7, 9, 9]
n = len(s) - 1

selected_activities = [0] + recursive_activity_selector(s, f, 0, n)
print("Selected activity indices:", selected_activities)
```
